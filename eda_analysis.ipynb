```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
from pathlib import Path

# Set visual style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 150
plt.rcParams["font.sans-serif"] = "DejaVu Sans"

print("Environment initialized successfully!")
```

### 1. Ingestion & Initial Inspection of `dataset.csv`

```python
df = pd.read_csv('dataset.csv')
print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns\n")
print("Column Info & Null Value Summary:")
print(df.info())
display(df.head(10))
```

### 2. Descriptive Statistics

```python
display(df.describe(include='all'))
```

### 3. Visualisation 1: Crop Species & Pathological Label Distribution
Analyzes the frequency of inspection records across the 5 botanical crops (Apple, Corn, Pepper, Potato, Tomato) and 5 pathological conditions.

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.countplot(data=df, x='crop_type', hue='disease_label', palette='Set2', ax=ax)
ax.set_title('Pathological Disease Label Distribution Across Crop Types (N=500)', fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Crop Type', fontsize=11, fontweight='bold')
ax.set_ylabel('Number of Leaf Samples', fontsize=11, fontweight='bold')
ax.legend(title='Disease Label', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### 4. Visualisation 2: Confidence Score Density by Disease Label
Evaluates model certainty distributions to ensure confidence scores fall within acceptable bounds [0.701, 0.999].

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.kdeplot(data=df, x='confidence_score', hue='disease_label', common_norm=False, fill=True, alpha=0.3, ax=ax)
ax.set_title('Prediction Confidence Probability Density Across Disease Categories', fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Model Confidence Score (0.0 to 1.0)', fontsize=11, fontweight='bold')
ax.set_ylabel('Density', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5. Visualisation 3: Physical Leaf Surface Area vs. Pathological Severity

```python
fig, ax = plt.subplots(figsize=(10, 5))
severity_order = ['None', 'Mild', 'Moderate', 'Severe']
sns.boxplot(data=df, x='severity', y='leaf_area_cm2', order=severity_order, palette='YlGnBu', ax=ax)
ax.set_title('Physical Leaf Blade Surface Area (cm²) Grouped by Infection Severity', fontsize=13, fontweight='bold', pad=12)
ax.set_xlabel('Pathological Severity Rating', fontsize=11, fontweight='bold')
ax.set_ylabel('Leaf Area (cm²)', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 6. Visualisation 4: Exploratory Data Quality Audit & Anomaly Detection
Identifies records where `disease_label == 'Healthy'` yet `severity != 'None'`.

```python
healthy_df = df[df['disease_label'] == 'Healthy']
contradiction_counts = healthy_df['severity'].value_counts()
print("Breakdown of Severity Ratings in 'Healthy' Samples:")
print(contradiction_counts)

fig, ax = plt.subplots(figsize=(8, 4.5))
contradiction_counts.plot(kind='bar', color=['#2ecc71', '#f39c12', '#e67e22', '#e74c3c'], ax=ax, edgecolor='black')
ax.set_title("Data Hygiene Audit: Contradictory Severity Ratings on 'Healthy' Leaves", fontsize=12, fontweight='bold', pad=12)
ax.set_xlabel('Tagged Severity Value', fontsize=10, fontweight='bold')
ax.set_ylabel('Record Count', fontsize=10, fontweight='bold')
for p in ax.patches:
    ax.annotate(str(p.get_height()), (p.get_x() + p.get_width() / 2., p.get_height() / 2), ha='center', va='center', color='white', fontweight='bold')
plt.tight_layout()
plt.show()
```
